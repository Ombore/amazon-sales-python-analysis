***Question 1: What is the total revenue?***

```python
total_revenue = df["total_revenue"].sum()
print("Total Revenue:", total_revenue)
```

***Question 2: What is the average revenue?***

```python
average_revenue = df["total_revenue"].mean()
print("Average Revenue:", average_revenue)
```

***Question 3: Which region generates the most revenue?***

```python
region_revenue = (
    df.groupby("customer_region")["total_revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(region_revenue)
```

```python
# Average and Median Revenue by region (EXTRA)
avg_med_revenue = (
    df.groupby("customer_region")["total_revenue"]
      .agg(["mean", "median"])
      .sort_values(by="mean", ascending=False)
)

print(region_revenue)
```

***`Visualization`***

```python
region_revenue.plot(kind="bar")

plt.title("Revenue by Customer Region")
plt.xlabel("Customer Region")
plt.ylabel("Total Revenue")
plt.ticklabel_format(style="plain", axis="y")
plt.xticks(rotation=45)
plt.show()
```

***Which Region Has the Most Products?***

```python
region_products = (
    df.groupby("customer_region")["product_id"]
      .nunique()
      .sort_values(ascending=False)
)

print(region_products)
```

***Visualization***

```python
region_products.plot(kind="bar")

plt.title("Number of Products by Region")
plt.xlabel("Customer Region")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.show()
```

***Which Payment Method Is Most Popular?***

```python
payment_counts = df["payment_method"].value_counts()

print(payment_counts)
```

***Visualization***

```python
payment_counts.plot(kind="bar")

plt.title("Orders by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

***Average Discounted Price by Region***

```python
avg_price_region = (
    df.groupby("customer_region")["discounted_price"]
      .mean()
      .sort_values(ascending=False)
)

print(avg_price_region)
```

***Visualization***

```python
avg_price_region.plot(kind="bar")

plt.title("Average Discounted Price by Region")
plt.xlabel("Customer Region")
plt.ylabel("Average Discounted Price")
plt.xticks(rotation=45)
plt.show()
```

***Which Products Generate the Most Revenue?***

```python
product_revenue = (
    df.groupby("product_id")["total_revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(product_revenue.head(10))
```

***Visualization***

```python
top_products = product_revenue.head(10)

top_products.plot(kind="bar")

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product ID")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["total_revenue"], bins=30, kde=True)

plt.title("Revenue Distribution")
plt.xlabel("Total Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["discounted_price"], bins=30, kde=True)

plt.title("Discounted Price Distribution")
plt.xlabel("Discounted Price")
plt.ylabel("Frequency")

plt.show()
```

***Relationship Between Price and Revenue***

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="discounted_price",
    y="total_revenue"
)

plt.title("Discounted Price vs Total Revenue")
plt.xlabel("Discounted Price")
plt.ylabel("Total Revenue")

plt.show()
```

```python
correlation = df[
    ["discounted_price", "total_revenue"]
].corr()

print(correlation)
```

***Revenue by Payment Method***

```python
payment_revenue = (
    df.groupby("payment_method")["total_revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(payment_revenue)
```

***Visualization***

```python
payment_revenue.plot(kind="bar")

plt.title("Revenue by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Revenue")
plt.ticklabel_format(style="plain", axis="y")
plt.xticks(rotation=45)

plt.show()
```

***Top 10 Revenue-Generating Regions/Products***

```python
top_10_products = (
    df.groupby("product_id")["total_revenue"]
      .sum()
      .nlargest(10)
)

print(top_10_products)
```

***Find High-Revenue Transactions***

```python
high_revenue = df[
    df["total_revenue"] > 1000
]

print(high_revenue)
```

***Revenue Categories***

```python
df["revenue_category"] = pd.cut(
    df["total_revenue"],
    bins=[0, 500, 1000, 5000],
    labels=["Low", "Medium", "High"]
)
```

```python
df["revenue_category"].value_counts()
```

***Visualization***

```python
df["revenue_category"].value_counts().plot(kind="bar")

plt.title("Revenue Category Distribution")
plt.xlabel("Revenue Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)

plt.show()
```

***Correlation Analysis***

```python
numeric_columns = df.select_dtypes(
    include=np.number
)
```

***Visualization***

```python
correlation_matrix = numeric_columns.corr()

plt.figure(figsize=(10,7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.show()
```
